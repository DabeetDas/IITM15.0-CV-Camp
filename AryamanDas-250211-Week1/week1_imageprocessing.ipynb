```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
from google.colab.patches import cv2_imshow
```

```python
path = 'scott.jpg'  # Replace with your image path
img = cv2.imread(path, cv2.IMREAD_GRAYSCALE)
img = cv2.resize(img, None, fx=0.3, fy=0.3, interpolation=cv2.INTER_AREA)
cv2_imshow(img)
```

```python
img = cv2.GaussianBlur(img, (3, 3), 0)
cv2_imshow(img)
```

```python
sobel_x = np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]])
sobel_y = np.array([[-1, -2, -1], [0, 0, 0], [1, 2, 1]])
```

```python
gx = cv2.filter2D(img, cv2.CV_64F, sobel_x)
gy = cv2.filter2D(img, cv2.CV_64F, sobel_y)
```

```python
mag = np.sqrt(gx**2 + gy**2)
magn = np.uint8(255 * mag / np.max(mag))
```

```python
plt.imshow(magn, cmap='gray')
plt.axis('off')
```
