#Using predefined kernel Prewitt Filter

```python
import torch
import torch.nn as nn
from PIL import Image
import matplotlib.pyplot as plt

# Read image and convert to grayscale
image = Image.open("img.jpg").convert("L")

image_tensor = torch.tensor(
    list(image.getdata()),
    dtype=torch.float32
).reshape(1, 1, image.height, image.width)

# Prewitt X
prewitt_x = torch.tensor([
    [-1, 0, 1],
    [-1, 0, 1],
    [-1, 0, 1]
], dtype=torch.float32)

# Prewitt Y
prewitt_y = torch.tensor([
    [-1, -1, -1],
    [ 0,  0,  0],
    [ 1,  1,  1]
], dtype=torch.float32)

# Create convolution layer
conv = nn.Conv2d(
    in_channels=1,
    out_channels=1,
    kernel_size=3,
    stride=1,
    padding=0,
    bias=False
)

# Apply X filter
conv.weight.data = prewitt_x.reshape(1, 1, 3, 3)
conv.weight.requires_grad = False

output_x = conv(image_tensor)

# Apply Y filter
conv.weight.data = prewitt_y.reshape(1, 1, 3, 3)
conv.weight.requires_grad = False

output_y = conv(image_tensor)

# Combine X and Y gradients
edge_magnitude = torch.sqrt(output_x**2 + output_y**2)

# Display final edge image
plt.imshow(
    edge_magnitude.squeeze().detach().numpy(),
    cmap="gray"
)

plt.axis("off")
plt.show()
```
