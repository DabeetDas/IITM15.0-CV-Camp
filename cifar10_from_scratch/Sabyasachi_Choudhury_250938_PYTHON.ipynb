(Tip: Before starting this section, brush up on strings in python)

Write a function defintion that:
1. Adds the string "Hello, " to the beginning and "!" to the end of any given input.
2. Takes in two numbers, squares each number, then returns the sum of both squares.
3. Takes in value and returns the count of the nubmer of vowels in a sequence.
4. Takes in string and returns the string without any vowels.
5. Takes in a list of strings and returns the shortest string in the list.
6. Takes in sequence of numbers and returns the odd numbers in a list.
7. Takes in sequence of numbers and returns all the numbers added together.

```python
# your answer here
def func1(base):
  return f"Hello, {base}!"

def func2(a, b):
  return a**2 + b**2

def func3(seq):
  n = [1 for elem in seq.lower() if elem in "aeiou"]
  return len(n)

def func4(string):
  newstring = [c for c in string if c.lower() not in "aeiou"]
  return ''.join(newstring)

def func5(arr):
  n = len(arr[0])
  shortest = arr[0]
  for elem in arr:
    if len(elem) < n:
      n = len(elem)
      shortest = elem
  return shortest

def func6(arr):
  return [elem for elem in arr if elem & 1]

def func7(arr):
  return sum(arr)

#examples of each
print(func1("wourbg"))
print(func2(4, 5))
print(func3("oweugoeiwfp"))
print(func4("gangaaplle"))
print(func5(["apekrowebi", "apple", "sabyasachi", "Kritik", "BAnana"]))
print(func6([1, 34, 33, 12]))
print(func7([1, 2, 3, 4, 5, 6, 7]))
```

**Plots**

Plot the following plots using matplotlib, you many generate your own data or get it from the internet:

1. Histogram
2. Scatter Plot
3. Boxplot
4. Pie charts

(Bonus - Read up on the use-case of histograms in EDA and ML in general, they are VERY useful! Also, look up **Violin plots** if you have extra time)

```python
# your solution here

# Setup Subplot grid
fig = plt.figure(1, figsize=(8, 8))
((hist, scat), (box, pie)) = fig.subplots(2, 2)

# Histogram
hist_data = np.floor(np.random.normal(10, 4, size=(300,))).astype(np.int16)
hist_counts = {}
for elem in hist_data:
  if elem not in hist_counts:
    hist_counts[elem] = 1
  else:
    hist_counts[elem] += 1

hist.bar(hist_counts.keys(), hist_counts.values())

# Scatter
c1_data = [np.random.normal(10, 0.5, size=(50, )), np.random.normal(8, 1, size=(50, ))]
c2_data = [np.random.normal(4, 2, size=(50, )), np.random.normal(4, 2, size=(50, ))]
scat.scatter(c1_data[0], c1_data[1], color='red')
scat.scatter(c2_data[0], c2_data[1], color='blue')

#PIe chart
clubs = ["BCS", "FnA", "PClub", "AeroMod", "ICG", "Others"]
members = [100, 43, 135, 60, 30, 100]
pie.pie(members, labels=clubs, autopct="%.0f%%")

#Boxplot
mth = np.random.normal(20, 5, size=(100,))
phy = np.random.normal(50, 10, size=(100,))
lif = np.random.normal(80, 15, size=(100,))
box.set_ylabel("Score in Q1")
box.boxplot([mth, phy, lif], tick_labels=["mth", "phy", "lif"])
```
